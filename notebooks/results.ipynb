```python
# Set the project path and load pandas for the result tables.
from pathlib import Path
import pandas as pd

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
```

```python
# Combine per-seed summaries and display pilot and main experiment counts and forgetting rates.
for mode in ["pilot", "main"]:
    paths = sorted((ROOT / "results" / mode).glob("seed_*/summary.csv"))
    print(mode)
    if paths: display(pd.concat([pd.read_csv(p) for p in paths], ignore_index=True))
    else: print("No completed runs yet.")
```

```python
# Display saved predictor scores, correlations, and summaries across seeds.
for name in ["predictor_scores.csv", "associations.csv", "seed_summary.csv"]:
    path = ROOT / "results/tables" / name
    if path.exists():
        print(name)
        display(pd.read_csv(path))
```
